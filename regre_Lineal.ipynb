```python
#importamos librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df= pd.DataFrame({
    "horas":[1,2,3,4,5],
    "nota":[2,3,5,4,6]
})
x=df[["horas"]]
y=df[["nota"]]
print(x)
```

```python
#Creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(x,y)
```

```python
#Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
#Realizamos la prediccion
prediccion = modelo.predict([[6]])
print("Predicción: ", prediccion)
```

### Graficar con matplolib el modelo y verificar si el nuevo valor corresponde a la predicción en la recta.


```python
X_linea = np.linspace(1, 6, 100)
X_linea_df = pd.DataFrame({"horas": X_linea})

#predecimos los valores de 'nota' para cada punto de la recta
y_linea = modelo.predict(X_linea_df)

plt.figure(figsize=(8, 5))

#puntos de entrenamiento originales
plt.scatter(df["horas"], df["nota"], color="blue", label="Datos reales", s=60)
plt.plot(X_linea, y_linea, color="red", label="Recta de regresión", linewidth=2)
#valor predicho (6 horas, 6.7 nota)
plt.scatter([6], prediccion, color="green", s=120, zorder=5, label="Predicción (6h, 6.7)")

#configuración visual
plt.title("Regresión Lineal: Horas vs Nota")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```
