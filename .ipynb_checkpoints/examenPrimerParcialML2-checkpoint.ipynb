```python
#importacion de librerias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import opendatasets as od
from sklearn import metrics
from sklearn.model_selection import train_test_split, KFold, cross_validate
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.linear_model import LinearRegression
```

### CARGA Y COMPRENSION DE LOS DATOS

```python
dataset_url = "https://www.kaggle.com/datasets/mirichoi0218/insurance"
od.download(dataset_url)
```

```python
df = pd.read_csv("insurance/insurance.csv")
print("Dimensiones: ", df.shape)
df.head()
```

### EDA Y CALIDAD DE LOS DATOS

```python
#comprobación de valores ausentes y duplicados
print("Valores ausentes por columna:\n", df.isnull().sum())
print("\nRegistros duplicados:", df.duplicated().sum())

#eliminamos duplicados si existen para asegurar calidad
df = df.drop_duplicates().reset_index(drop=True)

#resumen estadístico
print("\nEstadísticas numéricas:")
display(df.describe())

#visualizamos de la distribución del target
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.histplot(df['charges'], kde=True, color='teal')
plt.title('Distribución de Cargos Médicos')

plt.subplot(1, 2, 2)
sns.boxplot(x='smoker', y='charges', data=df, palette='Set2')
plt.title('Cargos Médicos por Hábito de Fumar')
plt.tight_layout()
plt.show()
```

### DETECCIÓN DE OUTLIERS

```python
def detectar_outliers_iqr(df_in, columna):
    Q1 = df_in[columna].quantile(0.25)
    Q3 = df_in[columna].quantile(0.75)
    RIC = Q3 - Q1
    limite_inf = Q1 - 1.5 * RIC
    limite_sup = Q3 + 1.5 * RIC
    indices_outliers = df_in[(df_in[columna] < limite_inf) | (df_in[columna] > limite_sup)].index
    return indices_outliers, limite_inf, limite_sup

# Identificación de outliers en 'bmi'
outliers_bmi, li, ls = detectar_outliers_iqr(df, 'bmi')
print(f"Límites para BMI: [{li:.2f}, {ls:.2f}]")
print(f"Total de outliers detectados en 'bmi': {len(outliers_bmi)}")

# Para no perder representatividad clínica, se aplica recorte (capping/winsorization)
df['bmi'] = np.clip(df['bmi'], li, ls)
```

### PREPARACIÓN DE VARIABLES

```python
# Binarización de variables dicotómicas
df['sex'] = df['sex'].map({'female': 0, 'male': 1})
df['smoker'] = df['smoker'].map({'no': 0, 'yes': 1})

# One-Hot Encoding para 'region' (evitando multicolinealidad con drop_first=True)
df = pd.get_dummies(df, columns=['region'], drop_first=True, dtype=int)

# Separación de características independientes (X) y dependiente (y)
X = df.drop(columns=['charges'])
y = df['charges']

print("Características codificadas:\n", X.columns.tolist())
```

### NORMALIZACIÓN O ESTANDARIZACIÓN

```python
#división en particiones de train y test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

#estandarización de las características numéricas
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

#convertimos nuevamente a DataFrame con nombres de columnas
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X.columns)
```

### SELECCIÓN DE ATRIBUTOS

```python
selector = SelectKBest(score_func=f_regression, k=5)
X_train_sel = selector.fit_transform(X_train_scaled, y_train)
X_test_sel = selector.transform(X_test_scaled)

#visualizamos la importancia de las variables seleccionadas
importancias = pd.Series(selector.scores_, index=X.columns).sort_values(ascending=False)
print("Importancia estadística (F-score):\n", importancias)

columnas_seleccionadas = X.columns[selector.get_support()]
print("\nAtributos finales seleccionados:", columnas_seleccionadas.tolist())
```

### VALIDACIÓN DE DATOS

```python
modelo_cv = LinearRegression()
cv = KFold(n_splits=5, shuffle=True, random_state=42)

#métricas calculadas en las 5 iteraciones
scoring = ['neg_mean_absolute_error', 'r2']
cv_results = cross_validate(modelo_cv, X_train_sel, y_train, cv=cv, scoring=scoring)

mae_cv = -cv_results['test_neg_mean_absolute_error'].mean()
r2_cv = cv_results['test_r2'].mean()

print(f"MAE promedio en validación cruzada: {mae_cv:.4f}")
print(f"R² promedio en validación cruzada: {r2_cv:.4f}")
```

### ENTRENAMIENTO MEDIANTE REGRESIÓN LINEAL

```python
modelo_final = LinearRegression()
modelo_final.fit(X_train_sel, y_train)

#coeficientes obtenidos
print(f"Término independiente (Intercepto): {modelo_final.intercept_:.4f}")
for col, coef in zip(columnas_seleccionadas, modelo_final.coef_):
    print(f"Coeficiente para {col}: {coef:.4f}")
```

### EVALUACIÓN

```python
y_pred = modelo_final.predict(X_test_sel)

#calculamos las métricas
mae = metrics.mean_absolute_error(y_test, y_pred)
mse = metrics.mean_squared_error(y_test, y_pred)
rmse = metrics.root_mean_squared_error(y_test, y_pred)
mape = metrics.mean_absolute_percentage_error(y_test, y_pred)
r2 = metrics.r2_score(y_test, y_pred)

print(f"MAE:  {mae:.4f}")
print(f"MSE:  {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"MAPE: {mape:.4f}")
print(f"R²:   {r2:.4f}")

#gráfico de diagnóstico: valores Reales vs predicciones
plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred, alpha=0.6, color='royalblue', edgecolor='k')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2, label='Ajuste ideal')
plt.xlabel('Gastos Reales (charges)')
plt.ylabel('Gastos Predichos (charges)')
plt.title('Evaluación: Gastos Reales vs Predicción')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### PREDICCIÓN CON NUEVOS DATOS

```python
#nuevo registro de paciente
nuevo_paciente = pd.DataFrame([{
    'age': 35,
    'sex': 1,
    'bmi': 28.5,
    'children': 2,
    'smoker': 1,
    'region_northwest': 0,
    'region_southeast': 1,
    'region_southwest': 0
}])

#escalamos con el transformador ya entrenado
nuevo_paciente_scaled = pd.DataFrame(scaler.transform(nuevo_paciente), columns=X.columns)

#filtramos atributos seleccionados
nuevo_paciente_sel = selector.transform(nuevo_paciente_scaled)

#damos un estimo de costo médico
estimacion = modelo_final.predict(nuevo_paciente_sel)
print(f"Costo médico estimado para el nuevo paciente: ${estimacion[0]:,.2f}")
```
